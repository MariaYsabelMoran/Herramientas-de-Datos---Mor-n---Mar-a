```python
#Paso 3:
#Carga del dataset y además una primera vista de 10 columnas

Data_Set = pd.read_csv("/content/superbowl_commercials.csv")
Data_Set.head(10)
```

### Renombrar Columnas

```python
# Mostrar las columnas actuales para facilitar el mapeo
print("Columnas actuales:")
print(Data_Set.columns.tolist())

# Crear un diccionario para mapear los nombres de las columnas a español
column_mapping = {
    'Year': 'Año',
    'Brand': 'Marca',
    'Superbowl Ads Link': 'Enlace Anuncio Superbowl',
    'Youtube Link': 'Enlace Youtube',
    'Funny': 'Divertido',
    'Shows Product Quickly': 'Muestra Producto Rapido',
    'Patriotic': 'Patriótico',
    'Celebrity': 'Celebridad',
    'Danger': 'Peligro',
    'Animals': 'Animales',
    'Uses Sex': 'Usa Sexo',
    'Length': 'Duración',
    'Estimated Cost': 'Costo Estimado',
    'Youtube Views': 'Vistas Youtube',
    'Youtube Likes': 'Me Gusta Youtube',
    'TV Viewers': 'Espectadores TV'
}

# Renombrar las columnas
Data_Set = Data_Set.rename(columns=column_mapping)

print("\nColumnas renombradas correctamente. Mostrando las primeras 5 filas con los nuevos nombres:")
display(Data_Set.head())
```

##Problemas a solucionar
1. Todas las columnas tienen 249 mov pero la columna 13 y 14, tienen menos
2. La columna año tiene decimales
3. Todas las columnas parecieran tener mas de dos decimales
4. La columna de vistas en Youtobe es un numero muy grande, dificil de manejar

### Solución

1.  **Valores Nulos**: Se van a rellenar los valores nulos en 'Vistas Youtube' y 'Me Gusta Youtube' con 0 y luego convertir a tipo entero. También, se va a rellenar los valores nulos en 'Enlace Youtube' con cadenas vacías.
2.  **Decimales en Columnas Numéricas**: Redondear 'Costo Estimado' y 'Espectadores TV' a dos decimales.
3.  **Grandes Números**: Crear nuevas columnas para 'Vistas Youtube' (en millones) y 'Me Gusta Youtube' (en miles) para una mejor visualización.

```python
# 1. Manejar valores nulos
Data_Set['Vistas Youtube'] = Data_Set['Vistas Youtube'].fillna(0).astype(int)
Data_Set['Me Gusta Youtube'] = Data_Set['Me Gusta Youtube'].fillna(0).astype(int)
Data_Set['Enlace Youtube'] = Data_Set['Enlace Youtube'].fillna('')

# 2. Redondear columnas float a dos decimales
Data_Set['Costo Estimado'] = Data_Set['Costo Estimado'].round(2)
Data_Set['Espectadores TV'] = Data_Set['Espectadores TV'].round(2)

# 3. Crear nuevas columnas para Vistas Youtube y Me Gusta Youtube en unidades más manejables
Data_Set['Vistas Youtube (millones)'] = (Data_Set['Vistas Youtube'] / 1_000_000).round(2)
Data_Set['Me Gusta Youtube (miles)'] = (Data_Set['Me Gusta Youtube'] / 1_000).round(2)

print("Datos procesados. Mostrando la información actualizada y las primeras filas:")
Data_Set.info()
display(Data_Set.head())

print("\nVerificación final de valores nulos:")
display(Data_Set.isnull().sum())
```

### Verificación de Rangos Anómalos

Para identificar rangos anómalos o valores atípicos, vamos a utilizar las estadísticas descriptivas y visualizaciones. Esto nos permitirá examinar la distribución de los datos y detectar cualquier desviación significativa.

```python
# 1. Estadísticas Descriptivas para Columnas Numéricas
print("Estadísticas descriptivas de las columnas numéricas:")
display(Data_Set[['Año', 'Duración', 'Costo Estimado', 'Vistas Youtube (millones)', 'Me Gusta Youtube (miles)', 'Espectadores TV']].describe())

# 2. Visualización de Distribuciones (Histogramas y Box Plots)
# Definir las columnas numéricas a analizar
numeric_cols = ['Año', 'Duración', 'Costo Estimado', 'Vistas Youtube (millones)', 'Me Gusta Youtube (miles)', 'Espectadores TV']

plt.figure(figsize=(15, 10))
for i, col in enumerate(numeric_cols):
    plt.subplot(2, 3, i + 1) # Crear una cuadrícula de 2 filas y 3 columnas
    sns.histplot(Data_Set[col], kde=True)
    plt.title(f'Distribución de {col}')
    plt.xlabel(col)
    plt.ylabel('Frecuencia')
plt.tight_layout()
plt.show()

plt.figure(figsize=(15, 10))
for i, col in enumerate(numeric_cols):
    plt.subplot(2, 3, i + 1)
    sns.boxplot(y=Data_Set[col])
    plt.title(f'Box Plot de {col}')
    plt.ylabel(col)
plt.tight_layout()
plt.show()
```

### Pregunta 2: ¿Cómo han evolucionado las diferentes características de los anuncios a lo largo del tiempo?

Para esta pregunta, analizaremos la evolución de la duración, el costo estimado y la presencia de ciertas características (Divertido, Animales, Celebridad, etc.) a lo largo de los años.

```python
yearly_trends = Data_Set.groupby('Año').agg(
    avg_duration=('Duración', 'mean'),
    avg_cost=('Costo Estimado', 'mean'),
    prop_funny=('Divertido', lambda x: x.sum() / len(x) * 100),
    prop_animals=('Animales', lambda x: x.sum() / len(x) * 100),
    prop_celebrity=('Celebridad', lambda x: x.sum() / len(x) * 100),
    count_ads=('Año', 'count') # Para ver si el número de ads por año afecta la tendencia
).reset_index()

print("Tendencias anuales de características de anuncios:")
display(yearly_trends.head())

fig, axes = plt.subplots(3, 1, figsize=(12, 18))

sns.lineplot(ax=axes[0], x='Año', y='avg_duration', data=yearly_trends)
axes[0].set_title('Evolución de la Duración Promedio de los Anuncios')
axes[0].set_ylabel('Duración Promedio (segundos)')

sns.lineplot(ax=axes[1], x='Año', y='avg_cost', data=yearly_trends)
axes[1].set_title('Evolución del Costo Estimado Promedio de los Anuncios')
axes[1].set_ylabel('Costo Promedio (millones USD)')

sns.lineplot(ax=axes[2], x='Año', y='prop_funny', label='Divertido', data=yearly_trends)
sns.lineplot(ax=axes[2], x='Año', y='prop_animals', label='Animales', data=yearly_trends)
sns.lineplot(ax=axes[2], x='Año', y='prop_celebrity', label='Celebridad', data=yearly_trends)
axes[2].set_title('Evolución de la Proporción de Anuncios con Características Específicas')
axes[2].set_ylabel('Proporción (%)')
axes[2].legend()

plt.tight_layout()
plt.show()
```

### Pregunta 3: ¿Qué características tienen los anuncios que generan mayor impacto o mejor valoración por parte de la audiencia?

Para esta pregunta, utilizaremos 'Vistas Youtube (millones)' y 'Me Gusta Youtube (miles)' como métricas de impacto y analizaremos su relación con las características binarias de los anuncios.

```python
impact_metrics = ['Vistas Youtube (millones)', 'Me Gusta Youtube (miles)']
boolean_features = ['Divertido', 'Muestra Producto Rapido', 'Patriótico', 'Celebridad', 'Peligro', 'Animales', 'Usa Sexo']

plt.figure(figsize=(15, 10))

for i, metric in enumerate(impact_metrics):
    plt.subplot(1, 2, i + 1)
    avg_impact = []
    for feature in boolean_features:
        avg_true = Data_Set[Data_Set[feature] == True][metric].mean()
        avg_false = Data_Set[Data_Set[feature] == False][metric].mean()
        avg_impact.append({'Característica': feature, 'Tiene Característica': 'Sí', 'Impacto Promedio': avg_true})
        avg_impact.append({'Característica': feature, 'Tiene Característica': 'No', 'Impacto Promedio': avg_false})

    df_avg_impact = pd.DataFrame(avg_impact)

    sns.barplot(x='Característica', y='Impacto Promedio', hue='Tiene Característica', data=df_avg_impact)
    plt.title(f'Impacto Promedio por Característica ({metric})')
    plt.xticks(rotation=45, ha='right')
    plt.ylabel(metric)

plt.tight_layout()
plt.show()
```
